```python
text = "The cat is good and the cat is cute"

text = text.lower()
words = text.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
print(sorted(frequency))
sorted_dict = {}
for key in sorted(frequency):
    sorted_dict[key] = frequency[key]
print(sorted_dict)
```
