```python
text = "The cat is good and the cat is cute"

text = text.lower()
words = text.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
print(sorted(frequency))
sorted_dict = {}
for key in sorted(frequency):
    sorted_dict[key] = frequency[key]
print(sorted_dict)
```

AND → & → dono 1 hone chahiye

OR → | → koi ek 1

XOR → ^ → different = 1

NOT → ~ → bits ulte

Left shift → << → × 2

Right shift → >> → ÷ 2

```python
a = 5
b = 3

print("AND :", a & b)
print("OR  :", a | b)
print("XOR :", a ^ b)
print("NOT :", ~a)
print("Left Shift :", a << 1)
print("Right Shift:", a >> 1)
```
