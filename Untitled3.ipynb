Write a function that accepts two positive integers, viz. a and b and returns a list
of all the even numbers between a and b (including a and not including b).
For example: Even numbers between 10 and 20. [10,12,14,16,18]

```python
def even(a,b):
  list1=[]
  for i in range(a,b+1):
    if i%2==0:
      list1.append(i)

  return list1


print(even(10,20))
```

Write a function is_Lst_Palindrome(Lst) to check whether a list is palindrome. It
should return True if Lst is palindrome and False if Lst is not palindrome.

```python
def Lst_palindrome(a):
  left=0
  right=len(a)-1
  while left<right:
    if a[left]!=a[right]:
      return False
    else:
      left+=1
      right-=1
  return True

list1=[1,2,3,1,2,1]
print(Lst_palindrome(list1))
```

10)Write a function check_duplicate(Lst) which returns True if a list Lst contains
duplicate elements. It should return False, if all the elements in the list Lst are
unique.
For example:
Lst = [4,6,2,1,6,7,4]
Should return true as 4 and 6 appears more than once.
Lst = [1,2,3,12,4]
Should return false as all the elements appears only once.

```python
def check_duplicate(lst):
  length=len(lst)-1

  for i in range(0,length):
    if lst.count(i)>1:
      return True

  return False


list1=[4,6,2,1,6,7,4]
list2=[1,2,3,12,4]
print(check_duplicate(list1))
print(check_duplicate(list2))
```
